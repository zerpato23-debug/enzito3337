```python
import sys

def leer_archivo(ruta):
    """Lee el archivo y devuelve una lista de enteros."""
    datos = []
    try:
        with open(ruta, 'r') as f:
            for linea in f:
                for valor in linea.split():
                    datos.append(int(valor))
    except FileNotFoundError:
        raise FileNotFoundError(f"No se encontró el archivo: {ruta}")
    except ValueError:
        raise ValueError("El archivo contiene datos que no son números enteros")
    return datos


def calcular_promedio(lista, n1, n2):
    """Calcula el promedio entre las posiciones n1 y n2 según las reglas dadas."""
    if n2 < n1:
        return 0
    if n1 > len(lista):
        return 0
    if n2 > len(lista):
        n2 = len(lista)

    sublista = lista[n1:n2]
    if not sublista:
        return 0
    return sum(sublista) / len(sublista)


def main():
    try:
        if len(sys.argv) != 4:
            raise ValueError("Uso: python proyecto1.py <archivo> <n1> <n2>")

        archivo = sys.argv[1]
        n1 = int(sys.argv[2])
        n2 = int(sys.argv[3])

        if n1 < 0 or n2 < 0:
            raise ValueError("Los dos números enteros deben ser mayores o iguales a 0")

        datos = leer_archivo(archivo)
        promedio = calcular_promedio(datos, n1, n2)
        print(f"Promedio entre posiciones {n1} y {n2}: {promedio}")

    except ValueError as e:
        print(f"Error: {e}")
    except FileNotFoundError as e:
        print(f"Error: {e}")
    except Exception as e:
        print(f"Error inesperado: {e}")


if __name__ == "__main__":
    main()
```
